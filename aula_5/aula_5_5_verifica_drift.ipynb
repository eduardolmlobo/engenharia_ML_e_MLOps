```python
# Verifica se há drift detectado na tabela de log
drift_count = spark.sql("""
    SELECT COUNT(*) as total_drift
    FROM workspace.default.log_monitoramento_drift
    WHERE drift_detectado = true
""").collect()[0]["total_drift"]

print(f"Total de drifts detectados: {drift_count}")

# Define o valor da task para o condition task avaliar
if drift_count > 0:
    dbutils.jobs.taskValues.set(key="drift_detectado", value="true")
    print("Drift detectado! Acionando re-treino...")
else:
    dbutils.jobs.taskValues.set(key="drift_detectado", value="false")
    print("Nenhum drift detectado. Re-treino não necessário.")
```
